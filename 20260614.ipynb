```python
def function(name,*args,**kwargs):
    print(name)
    print(args)
    print(kwargs)

function("spam","ham",kwargs1="eggs",kwarg2="spamhameggs")
```

```python
def conect(*args,sep="/"):
    return sep.join(args)

words = ["spam","ham","eggs"]
options = {"sep":"&"}
print(conect(*words,**options))
# conect("spam", "ham", "eggs", sep="&") これと同じ意味
```
