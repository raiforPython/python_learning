```python
def func(values):
    print(f"func内 - valuesのID: {id(values)}") # valuesのIDを表示
    values.append(100)
    return values

nums = [1, 2, 3]
print(f"呼び出し前 - numsのID: {id(nums)}") # numsのIDを表示
new_nums = func(nums)
print(f"呼び出し後 - numsのID: {id(nums)}") # 変更後のnumsのIDを表示
print(f"new_numsのID: {id(new_nums)}") # new_numsのIDを表示

print(nums)
print(new_nums)
```

```python
def func_with_shallow_copy(values):
    copied_values = values.copy()
    print(f"func内 - copied_valuesのID: {id(copied_values)}") # コピー後のIDを表示
    copied_values.append(100)
    return copied_values

nums = [1, 2, 3]
print(f"呼び出し前 - numsのID: {id(nums)}")
new_nums = func_with_shallow_copy(nums)
print(f"呼び出し後 - numsのID: {id(nums)}")
print(f"new_numsのID: {id(new_nums)}")

print(nums)
print(new_nums)
```

```python
original_list = [1, 2, 3] # 全てイミュータブルな数値
shallow_copied_list = original_list.copy()

print(f"元のリスト: {original_list}, ID: {id(original_list)}")
print(f"シャローコピー: {shallow_copied_list}, ID: {id(shallow_copied_list)}")

# 新しいリストの要素を変更してみる
shallow_copied_list[0] = 99

print(f"変更後 - 元のリスト: {original_list}")
print(f"変更後 - シャローコピー: {shallow_copied_list}")

# 結果: 元のリストは変更されていない
# original_list と shallow_copied_list は異なるIDを持つが、中の要素はイミュータブルなので問題ない
```

```python
original_list_nested = [1, [2, 3], 4] # 中央にミュータブルなリストを含む
shallow_copied_list_nested = original_list_nested.copy()

print(f"元のリスト: {original_list_nested}")
print(f"シャローコピー: {shallow_copied_list_nested}")

# 各要素のIDを比較してみる
print(f"original_list_nested[1]のID: {id(original_list_nested[1])}")
print(f"shallow_copied_list_nested[1]のID: {id(shallow_copied_list_nested[1])}")
# -> 両者は同じIDを持つはず！

# シャローコピーのネストされたリストを変更してみる
shallow_copied_list_nested[1].append(99) # ここでネストされたリストを変更

print(f"変更後 - 元のリスト: {original_list_nested}")
print(f"変更後 - シャローコピー: {shallow_copied_list_nested}")

# 結果: 元のリストのネストされた要素も変更されてしまっている！
```
