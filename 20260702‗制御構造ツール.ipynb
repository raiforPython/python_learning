```python
def func():
    try:
        for i in range(3):
            if i == 1:
                raise ValueError("error")
            print(i)
    except ValueError:
        print("X")
    else: # tryブロック内で例外が全く発生しなかった場合にのみ実行
        print("Y")

func()
```

```python
def gen():
    yield 1
    yield 2
    return 3

g = gen()

try:
    print(next(g))
    print(next(g))
    print(next(g))
except StopIteration as e:
    print("X", e.value)
```

```python
pairs = [(1, 2), (3, 4), (5, 6)]
result = 0

for a, b in pairs:
    if a + b > 7:
        result += a
    else:
        result += b

print(result)
```

### ループ処理を関数として定義する

同じ処理を何度も繰り返す場合や、コードの可読性を高めるために、特定の処理を関数としてまとめることがよくあります。
先ほどの`pairs`リストの処理を関数`calculate_sum_based_on_pairs`として定義してみましょう。

```python
def calculate_sum_based_on_pairs(data_pairs):
    result = 0
    for a, b in data_pairs:
        if a + b > 7:
            result += a
        else:
            result += b
    return result

# 関数の使い方をデモンストレーション
my_pairs = [(1, 2), (3, 4), (5, 6)]
final_sum = calculate_sum_based_on_pairs(my_pairs)
print(f"計算された合計値: {final_sum}")

# 別のデータで試すこともできます
another_pairs = [(1, 1), (5, 5), (2, 8)]
another_sum = calculate_sum_based_on_pairs(another_pairs)
print(f"別のデータでの合計値: {another_sum}")
```
