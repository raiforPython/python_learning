```python
x = (1, 2, 3)
y = x
x = x + (4,) #再割り当てを伴う連結操作
print(y)
```

```python
x = [1, 2, 3]
y = x
x.append(4) #同じリストオブジェクトを直接変更
print(y)
```

```python
a = [1, 2, 3]
b = a[1:3]
print(b)
```

```python
a = [1, 2, 3]
b = a[:]
a.append(4)#「参照渡し」 リストのコピーではない
print(b)
```

```python
s = {1, 2, 3}
t = s
s.add(4)
print(t)
```

```python
def h(x, y):
    return x is y
#Pythonの is 演算子は、二つの変数がメモリ上で全く同じオブジェクトを参照しているかどうかをチェック
a = [1, 2]
b = [1, 2]
print(h(a, b))
```
