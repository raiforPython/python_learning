```python
from datetime import date

today = date.today()
print(today.year, today.month, today.day)
```

```python
from datetime import datetime

now = datetime.now()
print(now.hour,now.minute,now.second)
```

```python
from datetime import date

d = date(2025, 1, 15)
print(d.isoformat()) #ISO 8601形式
```

```python
from datetime import datetime, timedelta

dt = datetime(2024, 5, 1, 12, 0, 0)
print(dt)
new_dt = dt + timedelta(days=2, hours=3)
print(new_dt)
```

```python
from datetime import datetime

now = datetime.now()
print(now.year,now.month,now.day,now.hour,now.minute,now.second)
```

```python
from datetime import datetime

dt = datetime(2024, 8, 10, 19, 45, 30)
print(dt.strftime("%Y/%m/%d %H:%M"))
```

```python
from datetime import datetime

dt = datetime(2024, 3, 15, 18, 20, 0)
print(dt.weekday())
```

```python
from datetime import datetime

dt = datetime(2024, 12, 1, 0, 0, 0)
print(dt.isoweekday())
```

```python
data = []
for i in [1,2,3]:
    for j in [1,2]:
        if i != j:
            data.append((i,j))
print(data)
```
