```python
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")

class C(B):
    def show(self):
        super().show()
        print("C")

c = C()
c.show()
```

```python
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        super().show()
        print("B")
        print("---")

class C(B):
    def show(self):
        super().show()
        print("C")

c = C()
c.show()
```

```python
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")

class C(A):
    def show(self):
        print("C")

class D(B, C):
    def show(self):
        super().show()
        print("D")

d = D()
d.show()
```

```python
class A:
    def __init__(self):
        self.x = 10

class B(A):
    def __init__(self):
        super().__init__()
        self.y = 20

class C(B):
    def __init__(self):
        super().__init__()
        self.x = 30

obj = C()
print(obj.x, obj.y)
```

```python
class A:
    def show(self):
        return "A"

class B(A):
    def show(self):
        return "B"

class C(B):
    pass

obj = C()
print(obj.show())
```

```python
class A:
    x = 10
    def __init__(self):
        self.x = 20

class B(A):
    pass

obj = B()
print(obj.x, A.x, B.x)
```

```python
class A:
    def __init__(self):
        self.value = "A"

    def get(self):
        return self.value

class B(A):
    def __init__(self):
        super().__init__()
        self.value = "B"

    def get(self):
        return "B:" + super().get()

class C(B):
    pass

obj = C()
print(obj.get())
```

```python
class A:
    msg = "A"
    def get(self):
        return self.msg

class B(A):
    msg = "B"

class C(B):
    def __init__(self):
        self.msg = "C"

obj = C()
print(obj.get())
print(A.msg, B.msg, C.msg)
```

```python
class A:
    def msg(self):
        return "A"

class B:
    def msg(self):
        return "B"

class C(A, B):
    pass

obj = C()
print(obj.msg())
```

```python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

# DクラスのMROを表示
print(D.__mro__)
```

出力は `(<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>)` のようになります。これは、`D`のインスタンスでメソッドが呼び出された場合、まず`D`自身を検索し、次に`B`、その次に`C`、そして`A`、最後に`object`クラスを検索するという順序を示しています。

`C(B, A)`の例のように、多重継承の親クラスの順序を変更すると、MROもそれに応じて変化します。

```python
class A:
    def __len__(self):
        return 5

a = A()
print(len(a))
```

```python
class A:
    def __getitem__(self, index):
        return index * 2

a = A()
print(a[3])
```

```python
class A:
    count = 10

    @classmethod
    def get_count(cls):
        return cls.count

class B(A):
    count = 20

print(A.get_count(), B.get_count())
```

```python
class A:
    @staticmethod
    def add(x, y):
        return x + y

print(A.add(3, 4))
```

```python
class A:
    count = 10

    @classmethod
    def inc(cls):
        cls.count += 1

class B(A):
    count = 20

A.inc()
B.inc()

print(A.count, B.count)
```
